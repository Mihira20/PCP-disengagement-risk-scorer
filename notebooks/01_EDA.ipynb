```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
bene_2008 = pd.read_csv('../data/raw/DE1_0_2008_Beneficiary_Summary_File_Sample_1.csv')
bene_2009 = pd.read_csv('../data/raw/DE1_0_2009_Beneficiary_Summary_File_Sample_1.csv')
bene_2010 = pd.read_csv('../data/raw/DE1_0_2010_Beneficiary_Summary_File_Sample_1.csv')
print(bene_2008.head())
print(bene_2009.head())
print(bene_2010.head())
```

```python
print(bene_2008.shape)
print(bene_2009.shape)
print(bene_2010.shape)
```

```python
#print(bene_2008.columns.tolist())
print(bene_2008['BENE_DEATH_DT'].value_counts(dropna=False))
death_percent = (bene_2008['BENE_DEATH_DT'].notna().sum()/bene_2008.shape[0])*100
print(f"The death percentage in the year 2008 is:{death_percent:.2f}%")
```

```python
merged_df = pd.merge(bene_2008, bene_2009, on='DESYNPUF_ID')
merged_df = pd.merge(merged_df, bene_2010, on='DESYNPUF_ID')
```

```python
print(merged_df.shape)
```

```python
# Find IDs present in all three years
ids_2008 = set(bene_2008['DESYNPUF_ID'])
ids_2009 = set(bene_2009['DESYNPUF_ID'])
ids_2010 = set(bene_2010['DESYNPUF_ID'])

common_ids = ids_2008.intersection(ids_2009).intersection(ids_2010)
print(f"Beneficiaries present in all three years: {len(common_ids)}")
```

```python
print(f"The percentage of memebers that survived all three years= {(len(common_ids)/len(bene_2008))*100:.2f} %")
```

```python
filtered_bene_2008=bene_2008.loc[bene_2008['DESYNPUF_ID'].isin(common_ids)]
filtered_bene_2009=bene_2009.loc[bene_2009['DESYNPUF_ID'].isin(common_ids)]
filtered_bene_2010=bene_2010.loc[bene_2010['DESYNPUF_ID'].isin(common_ids)]
print(filtered_bene_2008.shape)
print(filtered_bene_2009.shape)
print(filtered_bene_2010.shape)
```

```python
otp = pd.read_csv('../data/raw/DE1_0_2008_to_2010_Outpatient_Claims_Sample_1.csv', low_memory=False)
print(otp.shape)
```

```python
print(otp.columns.tolist())
```

```python
print(otp['CLM_FROM_DT'].dtype)
print(otp['CLM_FROM_DT'].head())
```

```python
otp = otp.dropna(subset=['CLM_FROM_DT'])

otp['CLM_FROM_DT'] = pd.to_datetime(
    otp['CLM_FROM_DT'].astype(int).astype(str),
    format='%Y%m%d'
)

print(otp['CLM_FROM_DT'].dtype)
print(otp['CLM_FROM_DT'].head())
print(f"Rows after dropping null dates: {otp.shape[0]}")
```

```python
#filtering to 2010 claims only
otp_2010 = otp[otp['CLM_FROM_DT'].dt.year == 2010]

#filtering to cohort members only
otp_2010_cohort = otp_2010[otp_2010['DESYNPUF_ID'].isin(common_ids)]

#finding members with at least one claim
mem_with_claims_2010 = set(otp_2010_cohort['DESYNPUF_ID'].unique())

#building my label in the filtered_bene_2010 for dienganged members.
filtered_bene_2010['DISENGAGED'] = filtered_bene_2010['DESYNPUF_ID'].isin(mem_with_claims_2010)
filtered_bene_2010['DISENGAGED'] = filtered_bene_2010['DISENGAGED'].apply(lambda x: 0 if x else 1)

print(filtered_bene_2010['DISENGAGED'].value_counts())
print(f"\nDisengagement rate: {filtered_bene_2010['DISENGAGED'].mean()*100:.2f}%")
```

```python
#overall data health
print(f"Total Cohort size: {len(filtered_bene_2010)}")
print(f"Engaged(0): {(filtered_bene_2010['DISENGAGED']==0).sum()}")
print(f"Disengaged(1): {(filtered_bene_2010['DISENGAGED']==1).sum()}")
print(f"Unique members with claims in 2010: {len(mem_with_claims_2010)}")
print(f"Unique memebers in out_patient 2010: {otp['DESYNPUF_ID'].nunique()}")
```
