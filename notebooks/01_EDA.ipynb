```python
# Find IDs present in all three years
ids_2008 = set(bene_2008['DESYNPUF_ID'])
ids_2009 = set(bene_2009['DESYNPUF_ID'])
ids_2010 = set(bene_2010['DESYNPUF_ID'])

common_ids = ids_2008.intersection(ids_2009).intersection(ids_2010)
print(f"Beneficiaries present in all three years: {len(common_ids)}")
```

```python
print(f"The percentage of memebers that survived all three years= {(len(common_ids)/len(bene_2008))*100:.2f} %")
```

```python
filtered_bene_2008=bene_2008.loc[bene_2008['DESYNPUF_ID'].isin(common_ids)]
filtered_bene_2009=bene_2009.loc[bene_2009['DESYNPUF_ID'].isin(common_ids)]
filtered_bene_2010=bene_2010.loc[bene_2010['DESYNPUF_ID'].isin(common_ids)]
print(filtered_bene_2008.shape)
print(filtered_bene_2009.shape)
print(filtered_bene_2010.shape)
```

```python
SP_cols=[col for col in filtered_bene_2009.columns if col.startswith('SP_') and col not in ['SP_STATE_CODE']]
print(SP_cols)
```

```python
chronic_features = filtered_bene_2009[['DESYNPUF_ID'] + SP_cols].copy()
```

```python
chronic_features['chronic_burden']=chronic_features[SP_cols].sum(axis=1)
print(chronic_features['chronic_burden'].describe())
```

```python
print(filtered_bene_2009[SP_cols].head(10))
print(filtered_bene_2009[SP_cols].describe())
```

```python
chronic_features[SP_cols] = chronic_features[SP_cols].replace({1: 1, 2: 0})
```

```python
chronic_features.head()
chronic_features['chronic_burden']=chronic_features[SP_cols].sum(axis=1)
print(chronic_features['chronic_burden'].describe())
```

```python
chronic_features.head()
```

```python
chronic_features.drop(columns="Total_chronic_count")
```

```python
# Get 2008 chronic burden
chronic_2008 = filtered_bene_2008[['DESYNPUF_ID'] + SP_cols].copy()
chronic_2008[SP_cols] = chronic_2008[SP_cols].replace({1: 1, 2: 0})
chronic_2008['chronic_burden_2008'] = chronic_2008[SP_cols].sum(axis=1)

# Merge and calculate trend
chronic_features = chronic_features.merge(
    chronic_2008[['DESYNPUF_ID', 'chronic_burden_2008']], 
    on='DESYNPUF_ID'
)
chronic_features['chronic_burden_trend'] = chronic_features['chronic_burden'] - chronic_features['chronic_burden_2008']

print(chronic_features['chronic_burden_trend'].describe())
print(chronic_features.head())
```
